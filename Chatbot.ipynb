```python
import os
import shutil

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma

PDF_PATH = "data/STSV2025_ONLINE.pdf"
DB_PATH = "chroma_db"

# Xóa database cũ nếu đã tạo lỗi giữa chừng
if os.path.exists(DB_PATH):
    shutil.rmtree(DB_PATH)

loader = PyPDFLoader(PDF_PATH)
documents = loader.load()

print("Số trang đọc được:", len(documents))

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1500,
    chunk_overlap=150
)

chunks = splitter.split_documents(documents)

print("Số đoạn sau khi chia:", len(chunks))

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

vector_db = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory=DB_PATH
)

print("Đã tạo vector database xong.")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma

DB_PATH = "chroma_db"

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

vector_db = Chroma(
    persist_directory=DB_PATH,
    embedding_function=embeddings
)

print("Đã load vector database.")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
import os

def ask_question(question):
    retriever = vector_db.as_retriever(
        search_kwargs={"k": 4}
    )

    docs = retriever.invoke(question)

    context = "\n\n".join([
        f"Nguồn: trang {doc.metadata.get('page', 'không rõ') + 1}\n{doc.page_content}"
        for doc in docs
    ])

    llm = ChatGoogleGenerativeAI(
        model="gemini-2.5-flash",
        google_api_key=os.getenv("GOOGLE_API_KEY"),
        temperature=0.2
    )

    prompt = f"""
Bạn là chatbot hỗ trợ sinh viên HCMUS.

Chỉ trả lời dựa trên phần tài liệu được cung cấp bên dưới.
Nếu tài liệu không có thông tin, hãy nói: "Không tìm thấy thông tin này trong sổ tay sinh viên."
Không được tự bịa quy định.
Trả lời bằng tiếng Việt, rõ ràng, ngắn gọn.
Cuối câu trả lời phải ghi nguồn trang nếu có.

TÀI LIỆU:
{context}

CÂU HỎI:
{question}
"""

    response = llm.invoke(prompt)

    return response.content, docs
```

```python
question = "phòng công tác sinh viên?"

answer, docs = ask_question(question)

print("CÂU HỎI:")
print(question)

print("\nTRẢ LỜI:")
print(answer)

print("\nCÁC ĐOẠN TÀI LIỆU ĐƯỢC DÙNG:")
for i, doc in enumerate(docs, start=1):
    page = doc.metadata.get("page", "không rõ")
    if isinstance(page, int):
        page += 1

    print("\n" + "=" * 80)
    print(f"Đoạn {i} - Trang {page}")
    print(doc.page_content[:1200])
```

```python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output

question_box = widgets.Text(
    placeholder="Nhập câu hỏi về sổ tay sinh viên...",
    description="Câu hỏi:",
    layout=widgets.Layout(width="80%")
)

ask_button = widgets.Button(
    description="Hỏi",
    button_style="primary"
)

output = widgets.Output()

def on_ask_clicked(b):
    with output:
        clear_output()

        question = question_box.value.strip()

        if not question:
            print("Nhập câu hỏi đi, chatbot không đọc được sự im lặng của mày đâu.")
            return

        print("Đang tìm câu trả lời...")

        answer, docs = ask_question(question)

        clear_output()
        display(Markdown("### Trả lời"))
        display(Markdown(answer))

        display(Markdown("### Nguồn truy xuất"))
        for i, doc in enumerate(docs, start=1):
            page = doc.metadata.get("page", "không rõ")
            if isinstance(page, int):
                page += 1

            display(Markdown(f"**Đoạn {i} - Trang {page}**"))
            display(Markdown(doc.page_content[:1000]))

ask_button.on_click(on_ask_clicked)

display(question_box, ask_button, output)
```
